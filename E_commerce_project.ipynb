**Which categories should the business investigate for freight-cost reduction?**

```python
category_analysis = (
    df.groupby('product_category_name')
    .agg(
        Sales=('price', 'sum'),
        Freight=('freight_value', 'sum'),
        Orders=('Orders_id', 'nunique')
    )
)

category_analysis['Freight_Burden_%'] = (
    category_analysis['Freight'] /
    category_analysis['Sales'] * 100
)

median_sales = category_analysis['Sales'].median()
median_freight_burden = category_analysis['Freight_Burden_%'].median()

investigate_categories = category_analysis[
    (category_analysis['Sales'] < median_sales) &
    (category_analysis['Freight_Burden_%'] > median_freight_burden)
]

print(
    investigate_categories
    .sort_values('Freight_Burden_%', ascending=False)
)
```
